```python
import ast
import pandas as pd
import d_pride_lfq_functions_proteins as lfq

master = pd.read_excel("b_coverage_reports/dataset_list.xlsx", dtype=str)


for _, row in master.iterrows():

    DATASET = {
        "accession": row["PRIDE_accession"].strip().upper(),
        "ID": row["ID"].strip(),
    }

    rule = row["Rule"].strip()

    GROUPS = ast.literal_eval(
        rule if rule.startswith("{") else f"{{{rule}}}"
    )
    output_folder = Path("c&d_aggrigated_datasets")
    output_folder.mkdir(parents=True, exist_ok=True)
    
    SETTINGS = {
        "group_aggregation": "mean",
        "presence_threshold": 0.10,
        "treat_zero_as_missing": False,
        "append_to_existing": False,
        "report_file": (output_folder/
            f"{DATASET['ID']}_{DATASET['accession']}.csv"
        ),
    }

    try:
        downloaded_tables, raw_catalog = (
            lfq.download_lfq_tables(DATASET["accession"])
        )

        lfq_catalog = lfq.make_numbered_lfq_catalog(
            raw_catalog
        )
        lfq_catalog.attrs.update(DATASET)

        result = lfq.build_range_representative_report(
            dataset=DATASET,
            groups=GROUPS,
            downloaded_tables=downloaded_tables,
            lfq_catalog=lfq_catalog,
            **SETTINGS,
        )

        print(f"✅ {DATASET['ID']} / {DATASET['accession']}")

    except Exception as error:
        print(
            f"⚠️ {DATASET['ID']} / {DATASET['accession']}: "
            f"{type(error).__name__}: {error}"
        )
```
